# Retail analysis

Exploration starts from project-relative data paths.

```python
from pathlib import Path

from retail_analysis.prepare_data import load_data, normalize_ids

customers, orders, order_items = load_data(Path.cwd())
customers['customer_id'] = normalize_ids(customers['customer_id'])
orders['customer_id'] = normalize_ids(orders['customer_id'])
```

```python
orders = orders[orders['status'] == 'completed']
orders = orders.dropna()
analysis = orders.merge(customers, on='customer_id').merge(order_items, on='order_id')
analysis['revenue'] = analysis['quantity'] * analysis['unit_price'] * (1 - analysis['discount'])
analysis['revenue'].sum()
```
